```python
# Step 2: Import APM and Scan a Coordinate
import src as apm

# Evaluate a prospective coordinate on the Saraswati-Ghaggar paleomeander
result = apm.scan(lat=29.5085, lon=74.8420)

print(f"Target Status: {'🌟 100% NEW / UNDISCOVERED LAND' if result.is_novel_undiscovered else '🏛️ KNOWN MONUMENT'}")
print(f"Calibrated Presence Probability: {result.calibrated_probability * 100:.1f}%")
print(f"Associated Cultural Horizon: {result.predicted_culture}")
print(f"Nearest Ancient Paleochannel: {result.nearest_paleochannel} ({result.dist_paleochannel_km:.2f} km)")
print(f"Distance to Nearest Monument: {result.dist_to_documented_monument_m / 1000.0:.2f} km from {result.nearest_documented_monument}")
```
